# 01. DATA LOADING & UNDERSTANDING

### Objective
Load the sales data from all three yearly sheets, combine them into one dataset, and understand its structure before starting the analysis.

### Dataset Details
- 2022 Sales Data
- 2023 Sales Data
- 2024 Sales Data

**Expected Output:** A single consolidated dataset for sales analysis.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Load Excel workbook
file_path = r"C:\Users\Roshni Pandey\Desktop\New folder\Data Analyst Interview Test - Answer\Data Analyst Interview Test - Answer\Python Assessment (Data Analyst)\Python Assessment (Data Analyst)\Python Assessment - Q6\3. Sample Sales Data.xlsx"

# Read yearly sheets
df_2022 = pd.read_excel(file_path, sheet_name="2022")
df_2023 = pd.read_excel(file_path, sheet_name="2023")
df_2024 = pd.read_excel(file_path, sheet_name="2024")

# Combine all yearly datasets
df = pd.concat(
    [df_2022, df_2023, df_2024],
    ignore_index=True
)

# Display results
print("Dataset loaded successfully!")
print("-" * 40)
print("2022 Records:", len(df_2022))
print("2023 Records:", len(df_2023))
print("2024 Records:", len(df_2024))
print("-" * 40)
print("Combined Dataset Shape:", df.shape)
print("Total Records:", df.shape[0])
print("Total Columns:", df.shape[1])
```

# 04. EXPLORATORY DATA ANALYSIS (EDA)

Exploratory Data Analysis (EDA) is the process of analyzing data to identify patterns, trends, relationships, and useful business insights.

In this section, we will analyze sales performance across different years, products, regions, sales channels, and customer types.

### Key Objectives
- Understand overall sales performance.
- Identify top-performing products and regions.
- Analyze yearly and monthly revenue trends.

## 4.1 Overall Sales Performance Analysis

**Business Question:** What is the overall sales performance of the business?

**Objective:** To evaluate total revenue, total units sold, total sales records, and average revenue per sales record.

```python
# Calculate Overall Sales Performance

sales_summary = pd.DataFrame({
    'Key Metrics': [
        'Total Revenue',
        'Total Units Sold',
        'Total Sales Records',
        'Average Revenue per Record'
    ],
    'Values': [
        round(df['Revenue'].sum(), 2),
        df['Units_Sold'].sum(),
        len(df),
        round(df['Revenue'].mean(), 2)
    ]
})

display(sales_summary)
```

## 4.2 Yearly Revenue Analysis

**Business Question:** Which year generated the highest revenue?

**Objective:** To compare annual revenue and identify the best-performing year.

**Business Value:** Helps the business evaluate its yearly sales performance and understand revenue trends.

```python
# Yearly Revenue Analysis

yearly_revenue = (
    df.groupby('Year')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=False)
)

yearly_revenue['Revenue'] = yearly_revenue['Revenue'].round(2)

display(yearly_revenue)

# Identify the highest revenue year
best_year = yearly_revenue.iloc[0]

print("Highest Revenue Year:", best_year['Year'])
print("Total Revenue:", round(best_year['Revenue'], 2))
```

## 4.3 Monthly Revenue Analysis

**Business Question:** Which month generated the highest revenue?

**Objective:** To analyze monthly revenue and identify the highest revenue-generating month.

**Business Value:** Helps the business identify seasonal sales patterns and understand which months perform better.

```python
# Monthly Revenue Analysis

monthly_revenue = (
    df.groupby('Month_Number')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=False)
)

# Add month names
monthly_revenue['Month'] = pd.to_datetime(
    monthly_revenue['Month_Number'], format='%m'
).dt.month_name()

monthly_revenue['Revenue'] = monthly_revenue['Revenue'].round(2)

display(monthly_revenue[['Month', 'Revenue']])

# Identify highest revenue month
best_month = monthly_revenue.iloc[0]

print("Highest Revenue Month:", best_month['Month'])
print("Total Revenue:", best_month['Revenue'])
```

## 4.4 Product Revenue Analysis

**Business Question:** Which product generated the highest revenue?

**Objective:** To compare the revenue generated by each product and identify the top-performing product.

**Business Value:** Helps the business identify its most valuable products and make better product and sales decisions.

```python
# Product Revenue Analysis

product_revenue = (
    df.groupby('Product')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=False)
)

product_revenue['Revenue'] = product_revenue['Revenue'].round(2)

display(product_revenue)

# Identify the highest revenue-generating product
top_product = product_revenue.iloc[0]

print("Top Performing Product:", top_product['Product'])
print("Total Revenue:", top_product['Revenue'])
```

## 4.13 Year-over-Year Revenue Growth Analysis

**Business Question:** How much did revenue increase or decrease compared to the previous year?

**Objective:** To calculate annual revenue growth and evaluate business performance over time.

**Business Value:** Helps the business understand whether revenue is growing or declining year over year.

```python
# Year-over-Year Revenue Growth Analysis

yearly_growth = (
    df.groupby('Year')['Revenue']
    .sum()
    .reset_index()
    .sort_values('Year')
)

yearly_growth['Revenue Growth (%)'] = (
    yearly_growth['Revenue'].pct_change() * 100
).round(2)

yearly_growth['Revenue'] = yearly_growth['Revenue'].round(2)

display(yearly_growth)

print("Year-over-Year Revenue Growth Analysis Completed!")
```

## 4.14 Product Revenue Contribution Analysis

**Business Question:** What percentage of total revenue is contributed by each product?

**Objective:** To calculate the revenue contribution percentage of each product.

**Business Value:** Helps the business identify which products contribute the most to overall revenue.

```python
# Product Revenue Contribution Analysis

product_contribution = (
    df.groupby('Product')['Revenue']
    .sum()
    .reset_index()
)

total_revenue = df['Revenue'].sum()

product_contribution['Revenue Contribution (%)'] = (
    product_contribution['Revenue'] / total_revenue * 100
).round(2)

product_contribution = product_contribution.sort_values(
    by='Revenue Contribution (%)',
    ascending=False
)

product_contribution['Revenue'] = product_contribution['Revenue'].round(2)

display(product_contribution)

print("Product Revenue Contribution Analysis Completed!")
```

## 4.15 Regional Revenue Contribution Analysis

**Business Question:** What percentage of total revenue is contributed by each region?

**Objective:** To calculate the revenue contribution percentage of each region.

**Business Value:** Helps the business understand the importance of different regional markets and identify key revenue-generating regions.

```python
# Regional Revenue Contribution Analysis

region_contribution = (
    df.groupby('Region')['Revenue']
    .sum()
    .reset_index()
)

total_revenue = df['Revenue'].sum()

region_contribution['Revenue Contribution (%)'] = (
    region_contribution['Revenue'] / total_revenue * 100
).round(2)

region_contribution = region_contribution.sort_values(
    by='Revenue Contribution (%)',
    ascending=False
)

region_contribution['Revenue'] = region_contribution['Revenue'].round(2)

display(region_contribution)

print("Regional Revenue Contribution Analysis Completed!")
```

## 5.1 Yearly Revenue Trend Analysis

**Business Question:** How did revenue change over the years 2022, 2023, and 2024?

**Chart Type:** Line Chart

**Objective:** To visualize yearly revenue trends and compare business performance across different years.

**Business Insight:** This visualization helps identify whether the business is experiencing revenue growth or decline over time.

```python
# Yearly Revenue Trend Analysis

yearly_revenue = (
    df.groupby('Year')['Revenue']
    .sum()
    .reset_index()
    .sort_values('Year')
)

plt.figure(figsize=(11, 5))

plt.plot(
    yearly_revenue['Year'],
    yearly_revenue['Revenue'],
    marker='o',
    linewidth=2.5,
    markersize=9
)

# Display revenue values
for x, y in zip(yearly_revenue['Year'], yearly_revenue['Revenue']):
    plt.annotate(
        f'{y:,.0f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 10),
        ha='center'
    )

plt.title('YEARLY REVENUE TREND (2022–2024)',
          fontsize=15, fontweight='bold', pad=15)

plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)

plt.xticks(yearly_revenue['Year'])
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

## 5.2 Monthly Revenue Trend Analysis

**Business Question:** Which month generates the highest revenue?

**Chart Type:** Line Chart

**Objective:** To analyze monthly revenue patterns and identify high-performing and low-performing months.

**Business Value:** Helps the business understand seasonal sales trends and improve monthly sales planning.

```python
# Monthly Revenue Trend Analysis

monthly_revenue = (
    df.groupby('Month_Number')['Revenue']
    .sum()
    .reset_index()
    .sort_values('Month_Number')
)

# Convert month numbers into month names
monthly_revenue['Month'] = pd.to_datetime(
    monthly_revenue['Month_Number'],
    format='%m'
).dt.month_name()

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_revenue['Month'],
    monthly_revenue['Revenue'],
    marker='o',
    linewidth=2.5,
    markersize=8
)

# Display revenue values
for x, y in zip(monthly_revenue['Month'], monthly_revenue['Revenue']):
    plt.annotate(
        f'{y:,.0f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 9),
        ha='center',
        fontsize=8
    )

plt.title('MONTHLY REVENUE TREND',
          fontsize=15, fontweight='bold', pad=15)

plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

## 5.5 Region-wise Revenue Comparison

**Objective:** To compare total revenue generated across different regions and identify the best-performing region.

**Business Value:** Helps the business identify strong regional markets and make better sales and marketing decisions.

```python
# Region-wise Revenue Comparison

region_revenue = (
    df.groupby('Region')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=False)
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    region_revenue['Region'],
    region_revenue['Revenue']
)

# Display revenue labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'{height:,.0f}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.title('REGION-WISE REVENUE COMPARISON',
          fontsize=15, fontweight='bold', pad=15)

plt.xlabel('Region', fontsize=12)
plt.ylabel('Total Revenue', fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

## 5.6 Sales Channel Revenue Contribution

**Business Question:** What percentage of total revenue is generated by each sales channel?

**Chart Type:** Pie Chart

**Objective:** To visualize the revenue contribution of different sales channels and compare their percentage shares.

**Business Value:** Helps the business understand which sales channels contribute the most revenue and where to focus marketing efforts.

```python
# Sales Channel Revenue Contribution

channel_revenue = (
    df.groupby('Channel')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 7))

plt.pie(
    channel_revenue,
    labels=channel_revenue.index,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.75,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 11}
)

plt.title(
    'SALES CHANNEL REVENUE CONTRIBUTION',
    fontsize=15,
    fontweight='bold',
    pad=20
)

plt.tight_layout()
plt.show()
```

## 5.8 Salesperson Performance Analysis

**Business Question:** How does each salesperson perform in terms of revenue generated?

**Chart Type:** Horizontal Bar Chart

**Objective:** To compare the total revenue generated by each salesperson and identify high-performing and low-performing salespeople.

**Business Value:** Helps the business evaluate sales team performance and identify areas where additional support or training may be useful.

```python
# Salesperson Performance Analysis

salesperson_revenue = (
    df.groupby('Salesperson')['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=True)
)

plt.figure(figsize=(10, 6))

bars = plt.barh(
    salesperson_revenue['Salesperson'],
    salesperson_revenue['Revenue']
)

# Display revenue labels
for bar in bars:
    width = bar.get_width()

    plt.text(
        width,
        bar.get_y() + bar.get_height() / 2,
        f' {width:,.0f}',
        va='center',
        fontsize=9
    )

plt.title(
    'SALESPERSON PERFORMANCE ANALYSIS',
    fontsize=15,
    fontweight='bold',
    pad=15
)

plt.xlabel('Total Revenue', fontsize=12)
plt.ylabel('Salesperson', fontsize=12)

plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
